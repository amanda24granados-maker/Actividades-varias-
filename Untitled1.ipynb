# Pregunta 1. Herencia y composición.

Problema de la herencia

En WhatsApp, una jerarquía incorrecta podría obligar a todas las funciones a tener un método para enviar mensajes, aunque algunas no lo necesiten.

```python
class FuncionWhatsApp:
    def enviar_mensaje(self):
        print("Enviando mensaje")

class Chat(FuncionWhatsApp):
    pass

class Estado(FuncionWhatsApp):
    pass

chat = Chat()
estado = Estado()

chat.enviar_mensaje()
estado.enviar_mensaje()
```

# Problema:
La clase Estado hereda un método que no necesita directamente. Esto demuestra que una herencia mal diseñada puede generar comportamientos innecesarios.


Solución usando composición:

```python
class EnviadorMensajes:
    def enviar(self, mensaje):
        print("Mensaje enviado:", mensaje)

class Chat:
    def __init__(self):
        self.enviador = EnviadorMensajes()

    def enviar_mensaje(self, mensaje):
        self.enviador.enviar(mensaje)

chat = Chat()
chat.enviar_mensaje("Hola, ¿cómo estás?")
```

Entonces la composición es preferible cuando una clase necesita utilizar una función sin heredar todos los métodos de otra. Mantendría la herencia cuando existe una relación clara entre clases y comparten comportamientos apropiados.

# Pregunta 2. Entrada, salida y excepciones.

En WhatsApp se pueden guardar conversaciones en un archivo. Es necesario controlar errores, como no encontrar el archivo o no tener permiso para escribir en él.

```python
class ErrorArchivo(Exception):
    pass

class ArchivoChats:
    def guardar(self, mensaje):
        try:
            with open("chats.txt", "a", encoding="utf-8") as archivo:
                archivo.write(mensaje + "\n")
        except OSError as error:
            raise ErrorArchivo("No se pudo guardar el mensaje") from error
        finally:
            print("Operación de guardado finalizada.")

    def leer(self):
        try:
            with open("chats.txt", "r", encoding="utf-8") as archivo:
                return archivo.read()
        except FileNotFoundError as error:
            raise ErrorArchivo("El archivo todavía no existe") from error
        except OSError as error:
            raise ErrorArchivo("No se pudo leer el archivo") from error

chats = ArchivoChats()

try:
    chats.guardar("Hola desde WhatsApp")
    print(chats.leer())
except ErrorArchivo as error:
    print("Error:", error)
```

Entonces el programa guarda y lee mensajes. Si el archivo no existe, no se puede leer o ocurre un problema al escribir, se comunica el error mediante una excepción propia. finally permite informar que terminó el intento de guardado, y with se encarga de cerrar el archivo.

# Pregunta 3. Polimorfismo y principios SOLID.

En esta miniaplicación, WhatsApp permite representar diferentes tipos de mensajes. Cada clase sobrescribe el método enviar(), demostrando polimorfismo.

```python
class Mensaje:
    def enviar(self):
        raise NotImplementedError("Debe implementarse en cada clase")

class MensajeTexto(Mensaje):
    def enviar(self):
        print("Enviando mensaje de texto")

class MensajeAudio(Mensaje):
    def enviar(self):
        print("Enviando mensaje de audio")

class AplicacionWhatsApp:
    def procesar(self, mensaje):
        mensaje.enviar()

app = AplicacionWhatsApp()

app.procesar(MensajeTexto())
app.procesar(MensajeAudio())
```

# Principios SOLID aplicados:

**1. Responsabilidad única** (SRP): cada clase tiene una función específica. Las clases de mensajes se encargan de su envío y la aplicación de procesarlos. Sin este principio, una sola clase podría encargarse de mensajes, archivos y usuarios, dificultando su mantenimiento.

2. **Abierto/cerrado (OCP):** se pueden agregar nuevos tipos de mensajes sin modificar la clase AplicacionWhatsApp. Sin este principio, habría que modificarla cada vez que se agregara un nuevo tipo de mensaje.

3. **Inversión de dependencias (DIP):**  la aplicación trabaja con el comportamiento general enviar() y no depende de una clase concreta. Sin este principio, la aplicación podría depender exclusivamente de MensajeTexto, dificultando incorporar audios u otros tipos de mensajes.


Estos ejemplos muestran que una buena organización del código permite reducir errores, reutilizar funciones y facilitar el mantenimiento, por lo que  la herencia es útil cuando existe una relación adecuada entre clases, mientras que la composición ofrece flexibilidad y SOLID ayuda a crear aplicaciones más organizadas y fáciles de ampliar.
